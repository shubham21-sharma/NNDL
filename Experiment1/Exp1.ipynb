```python
import numpy as np
X=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y=np.array([0,0,0,1])
w1=np.random.rand(1)
w2=np.random.rand(1)
w1_history=np.array([])
w2_history=np.array([])
b_history=np.array([])
b=np.zeros(1)
lr=0.1
epochs=10
def step_function(z):
    if z >0.5:
        return 1
    else:
        return 0
for epochs in range(epochs):
    for i in range(len(X)):
        a=w1 * X[i][0] + w2 * X[i][1] +b 
        y_pred=step_function(a)
        error= y[i] - y_pred
        w1+=lr*error*X[i][0]
        w2+=lr*error*X[i][1]
        b+=lr*error
print("Weight1",w1)
print("Weight2",w2)
print("Bias",b)
```
